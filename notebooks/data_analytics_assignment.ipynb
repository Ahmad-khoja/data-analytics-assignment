## Section 3 – Data Types Dictionary


```python
data = {
    "discrete": [1, 2, 3, 4, 5],
    "continuous": [1.2, 2.5, 3.8, 4.1, 5.0],
    "nominal": ["red", "green", "blue", "green", "red"],
    "ordinal": ["low", "medium", "medium", "high", "medium"]
}

data
```

## Section 4 – Statistics and Math


```python
import statistics
import math

# Mean for numerical data
mean_discrete = statistics.mean(data["discrete"])
mean_continuous = statistics.mean(data["continuous"])

# Mode for ordinal data
mode_ordinal = statistics.mode(data["ordinal"])

# Use a math function
sqrt_mean_discrete = math.sqrt(mean_discrete)

mean_discrete, mean_continuous, mode_ordinal, sqrt_mean_discrete
```

## Section 5 – Type Hinting and Reflection


```python
from typing import List
import statistics

def calculate_average(values: List[float]) -> float:
    """
    Calculate the mean value of a list of numbers.
    """
    return statistics.mean(values)

calculate_average(data["continuous"])
```
