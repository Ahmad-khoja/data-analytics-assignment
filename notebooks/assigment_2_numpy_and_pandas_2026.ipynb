```python
import pandas as pd
from pathlib import Path

data_path = Path("..") / "data" / "riistaonnettomuudet_2017_2020.csv"

riista_df = pd.read_csv(
    data_path,
    sep=',',
    encoding='latin1'
)

riista_df.head()


```

```python
accidents_by_month = (
    riista_df
    .groupby("kuukausi")
    .size()
    .sort_values(ascending=False)
)

accidents_by_month
```

#Comment#

The highest number of wildlife accidents occurs in late summer and autumn months.
There is a clear difference between months, with significantly more accidents during
these periods compared to winter months. This indicates strong seasonal variation
in wildlife movement and road crossings.


```python
cross_table = pd.crosstab(
    riista_df["riistalajiNimi"],
    riista_df["maakuntaNimi"]
)

cross_table
```

```python
cross_table_sorted = cross_table.loc[
    cross_table.sum(axis=1).sort_values(ascending=False).index
]

cross_table_sorted
```

#Comment#

The cross-tabulation shows that wildlife accidents are unevenly distributed
across species and regions. Certain species account for a large share of
accidents nationwide, while some regions experience significantly more
accidents than others. This reflects differences in wildlife populations
and traffic density between regions.

```python
riista_df["tapahtumaAika"] = (
    riista_df["tapahtumaAika"]
    .str.replace(r"\+.*$", "", regex=True)
)

riista_df["tapahtumaAika"] = pd.to_datetime(
    riista_df["tapahtumaAika"],
    errors="coerce"
)
```

```python
riista_df["tapahtumaAika"].dtype
```

```python
riista_df["hour"] = riista_df["tapahtumaAika"].dt.hour
riista_df["weekday"] = riista_df["tapahtumaAika"].dt.day_name()

riista_df[["tapahtumaAika", "hour", "weekday"]].head()

```

```python
accidents_by_hour = (
    riista_df
    .groupby("hour")
    .size()
    .sort_values(ascending=False)
)

accidents_by_hour
```

#Comment#

Wildlife accidents are most frequent during early morning and evening hours.
These times are particularly dangerous due to low visibility and increased
wildlife activity near roads.
